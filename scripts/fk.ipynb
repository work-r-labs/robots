```python
import numpy as np
import spatial_utils as su
from operator import matmul
from functools import reduce

from dataclasses import dataclass

np.set_printoptions(suppress=True)
```

```python
# crb15000_10kg = [
#     [su.rz_deg, 0, 0, 218],
#     [su.ry_deg, 150, 0, 182],
#     [su.ry_deg, 0, 0, 707],
#     [su.rz_deg, 0, 0, 110],
#     [su.ry_deg, 637, 0, 0],
#     [su.rz_deg, 101, 0, 80]
# ]

# j1, j2, j3, j4, j5, j6 = [0, 0, 0, 0, 0, 0]


IRB1200_5_90_STD = [
    [su.rz_deg, 0, 0, 0],
    [su.ry_deg, 0, 0, 399],
    [su.ry_deg, 0, 0, 448],
    [su.rz_deg, 451, 0, 42],
    [su.ry_deg, 0, 0, 0],
    # [su.rz_deg, 82, 0, 0]
]

IRB1200_5_90_STD_CASES = [
    # x y z        ax2, ax3
    ((451, 0, 889), (0, 0)),
    ((0, 0, 1300), (0, -85)),
    ((194, 0, 438), (0, 70)),
    ((901, 0, 402), (90, -85)),
    ((692, 0, -178), (130, -85)),
    ((-179, 0, -48), (-100, -200)),
    ((-72, 0, 583), (-100, 70)),
    ((-901, 0, 397), (-90, -85)),
    ((-887, 0, 240), (-100, -85)),
    ((458, 0, 549), (130, -200)),
]


def fk(spec, joints):
    assert len(spec) <= len(joints), f"{len(spec)} != {len(joints)}"
    sections = []
    for j, (axis, x, y, z) in zip(joints, spec):
        section = su.txyz(x, y, z) @ axis(j)
        sections.append(section)
    return reduce(matmul, sections)


for i, (expected_xyz, (ax2, ax3)) in enumerate(IRB1200_5_90_STD_CASES):
    joints_degrees = [0, ax2, ax3, 0, 0, 0]
    estimated_pose = fk(IRB1200_5_90_STD, joints_degrees)
    estimated_xyz = estimated_pose[:3, 3]

    if not np.allclose(estimated_xyz, expected_xyz, atol=1):
        # print()
        print(f"{i} FAILED. {expected_xyz} vs {tuple(estimated_xyz.round(4).tolist())}")
    else:
        # print()
        print(f"{i} passed")
```
